## **Question 1: Explain Normalization**

Normalization is the process of organizing data in a database to reduce redundancy and improve data integrity. The main goal is to ensure that data is stored logically, efficiently, and with minimal duplication. Normalization divides a large, unorganized table into smaller, well‑structured tables and establishes relationships between them.

### **Why do we normalize?**

* To remove duplicate data
* To ensure consistency across the database
* To make updates, inserts, and deletions more efficient
* To avoid anomalies (update, delete, insert anomalies)

### **Real‑World Example 1: Employee Records**

A company storing employee information in one big table may repeat department names for every employee. Normalization splits data into:

* Employee table
* Department table
  Every employee entry now references a department using a foreign key.

### **Real‑World Example 2: Online Orders**

An e‑commerce website storing all order details, customer details, and product details in one table leads to duplication. Normalization splits data into:

* Customers
* Products
* Orders
* OrderItems
  This reduces repetition and improves consistency.

---

## **Question 2: How to Implement Normalization?**

Implementing normalization involves applying a set of rules known as normal forms. The most commonly used ones are 1NF, 2NF, and 3NF.

### **Steps to Implement Normalization:**

1. Identify repeating groups → Apply **1NF**
2. Identify partial dependencies → Apply **2NF**
3. Identify transitive dependencies → Apply **3NF**
4. Create separate tables and use primary/foreign keys for relationships.

### **Real‑World Example 1: Student Database**

Suppose a table stores student, course, and instructor data. After normalization:

* Student table contains student info
* Course table contains course info
* Instructor table contains instructor info
* Enrollment table links student and course

### **Real‑World Example 2: Hospital Database**

A hospital may store patient, doctor, and appointment data in one table. Normalization splits it into:

* Patient
* Doctor
* Appointment
  Reduces storage and avoids inconsistency.

## **Question 3: What is Denormalization?**

Denormalization is the process of intentionally adding redundancy to improve read performance. It is used when normalized tables require expensive joins for read‑heavy systems.

### **Why do we denormalize?**

* To improve query performance
* To reduce the number of joins
* To optimize reporting and analytics queries

### **Real‑World Example 1: Reporting Database**

A data warehouse may store customer and order data in the same table to make report queries faster.

### **Real‑World Example 2: Social Media Timeline**

Instead of fetching posts by joining multiple tables, denormalized structures store pre‑joined data (user details + post details) for faster loading.

## **Question 5: Explain 1st, 2nd, and 3rd Normal Form**

Normalization is applied in stages known as normal forms. The first three are the most widely used.

---

### **1NF – First Normal Form**

A table is in **1NF** when:

* All values are **atomic** (no multiple values in a single column)
* No repeating groups

**Example Violation:** Phone column containing: `"98765, 98766"`

**Fix:** Put each phone number in a separate row or create a Phone table.

#### **Real-World Example 1: Customer Contacts**

If a customer has multiple phone numbers stored in one column, searching or filtering becomes difficult. Splitting them into separate rows makes data cleaner.

#### **Real-World Example 2: Course Enrollment**

A student enrolled in multiple courses must have each course in a separate row—not in a comma-separated list.

---

### **2NF – Second Normal Form**

A table is in **2NF** when:

* It is already in 1NF
* It has **no partial dependency** (i.e., a non-key column should not depend on part of a composite primary key)

**Fix:** Move partially dependent columns to a separate table.

#### **Real-World Example 1: Order Details**

Table with composite key (OrderID, ProductID) should not store attributes like ProductName—because ProductName depends only on ProductID.

#### **Real-World Example 2: Student-Course Mapping**

Student address stored in a table whose primary key is (StudentID, CourseID) violates 2NF since address depends only on StudentID.

---

### **3NF – Third Normal Form**

A table is in **3NF** when:

* It is already in 2NF
* It has **no transitive dependency** (non-key column must not depend on another non-key column)

**Fix:** Move dependent fields to their own table.

#### **Real-World Example 1: Employees**

If Employee table has DepartmentName, and DepartmentName depends on DepartmentID (another non-key field), move department data to Department table.

#### **Real-World Example 2: Product Inventory**

Storing SupplierAddress in Product table violates 3NF because SupplierAddress depends on SupplierID, not ProductID.

## **Question 6: Primary Key vs Unique Key**

Both constraints enforce uniqueness, but they differ in usage and behavior.

### **Primary Key**

* Uniquely identifies each row
* Only one per table
* Cannot contain NULL values

### **Unique Key**

* Ensures column values are unique
* Multiple unique keys allowed per table
* Can contain a single NULL (depends on DB engine)

#### **Real-World Example 1: Users Table**

* Primary Key → UserID
* Unique Key → Email (two users cannot share the same email)

#### **Real-World Example 2: Employee Table**

* Primary Key → EmployeeID

## **Question 7: Difference Between CHAR and VARCHAR**

### **CHAR**

* Fixed length
* Uses full defined space even if value is shorter
* Faster for fixed-size data

### **VARCHAR**

* Variable length
* Uses storage equal to actual data length
* More efficient when data sizes vary

#### **Real-World Example 1: Country Codes**

Country codes like 'IN', 'US', 'UK' are always 2 characters → Use CHAR(2).

#### **Real-World Example 2: Names**

Names vary in length → Use VARCHAR(50).

---

## **Question 8: Difference Between CHAR and NCHAR**

### **CHAR**

* Stores non-Unicode characters
* Uses 1 byte per character

### **NCHAR**

* Stores Unicode characters
* Uses 2 bytes per character
* Supports multiple languages and symbols

#### **Real-World Example 1: English-only Application**

If an app stores only English alphabets, CHAR is sufficient.

#### **Real-World Example 2: Multilingual Application**

Apps supporting Hindi, Chinese, Arabic, etc., must use NCHAR to store special characters.

---


## **Question 9: What’s the Size of CHAR vs NCHAR?**

### **CHAR:**

* Fixed-length
* Non‑Unicode
* **Storage:** 1 byte per character

### **NCHAR:**

* Fixed-length
* Unicode-supported
* **Storage:** 2 bytes per character

### Example:

* `CHAR(10)` always uses **10 bytes**
* `NCHAR(10)` always uses **20 bytes**

---

### **Real-World Example 1: Employee Code System**

If an organization uses fixed-size employee codes like `EMP001`, `EMP002` (6 characters), then `CHAR(6)` uses exactly 6 bytes.

But if the system needs to support multilingual alphabets (Hindi, Japanese, Arabic), then `NCHAR(6)` uses 12 bytes but stores international characters safely.

---

### **Real-World Example 2: Country Short Codes**

Country ISO codes (like "IN", "US") are always 2 characters.

* If only English alphabets are needed → use `CHAR(2)`
* If languages like Chinese are included → use `NCHAR(2)` (4 bytes)

## **Question 10: What Is the Use of an Index?**

Indexes are used to **speed up data retrieval** from a table. They act like a book’s index, allowing the database to locate data without scanning the entire table.

### **Why use an index?**

* Faster SELECT queries
* Better performance on large datasets
* Helps quickly locate rows based on search conditions

---

### **Real-World Example 1: Library Book Search**

A library index allows you to find a topic or chapter quickly.
Similarly, a SQL index helps find a row faster without reading every row.

---

### **Real-World Example 2: E‑commerce Product Search**

When a user searches for "iPhone 14" on an e‑commerce site, indexes on `ProductName` and `Category` make the search instant.
Without indexes, the site would scan millions of products → slow performance.

---

## **Question 11: How Does an Index Make Search Faster?**

Indexes work using **balanced trees (B‑tree)** or other structures like hash indexes. Instead of scanning every row, the database follows the tree to jump directly to the target record.

### **How it speeds up queries:**

* Reduces number of rows scanned
* Avoids full table scans
* Uses sorted, structured data for quick navigation

### **Technical Snapshot:**

* Without index → O(n) search time
* With B‑Tree index → O(log n)

---

### **Real-World Example 1: Phone Directory**

If searching for "Rahul" in a directory:

* Without index → read every name one by one
* With alphabetical ordering → jump to the section "R" directly

SQL index works the same way.

---

### **Real-World Example 2: Airport Boarding Pass Scanning**

Boarding passes have barcodes.

* Scanner jumps instantly to passenger record
* Without barcode, staff would manually search the passenger list (slow!)

Indexes play the role of barcodes in SQL tables.

---

## **Question 12: What Are the Two Types of Indexes?**

The two major types of indexes in SQL Server are:

### **1. Clustered Index**

* Sorts and stores the data rows physically
* Only **one** per table

### **2. Non‑Clustered Index**

* Separate structure from the data
* Stores pointers to actual data rows
* Multiple allowed per table

---

### **Real-World Example 1: Dictionary**

* A dictionary sorted alphabetically → **Clustered Index**
* Extra bookmark section you create (sticky notes on pages) → **Non‑Clustered Index**

---

### **Real-World Example 2: School Student Register**

* Roll numbers sorted by default → Clustered index
* Separate list of students sorted by name → Non‑clustered index pointing to roll numbers

---


## **Question 13: Clustered vs Non-Clustered Index**

Indexes come in two main types, and understanding the difference is essential for performance tuning.

---

### **Clustered Index**

* Physically **sorts and stores** the data in the table.
* One clustered index per table because data can be stored only in one order.
* Faster for range queries.

### **Non-Clustered Index**

* Stores a **separate index structure** that points to the actual data.
* Multiple non-clustered indexes can exist per table.
* Faster for point lookups.

---

### **Real-World Example 1: Dictionary**

* The actual dictionary pages are arranged alphabetically → **Clustered Index**.
* Sticky notes marking your favorite words → **Non-Clustered Index**.

---

### **Real-World Example 2: School Records**

* Student list sorted by roll number → Clustered
* Separate index sorted by student name → Non-clustered pointing to roll numbers.

## **Question 17: Differentiate Between AFTER Trigger vs INSTEAD OF Trigger**

Triggers execute automatically when a specific action happens on a table, but their purpose and timing differ.

---

### **AFTER Trigger**

* Executes **after** the actual INSERT, UPDATE, or DELETE operation occurs.
* Commonly used for logging, auditing, and cascading updates.
* If the main action fails, the AFTER trigger does not run.

### **INSTEAD OF Trigger**

* Executes **instead of** the triggering action.
* Allows overriding the default behavior.
* Useful when modifying views or preventing dangerous operations.

---

### **Real-World Example 1: Prevent Delete on Critical Table**

* Use an **INSTEAD OF DELETE** trigger to stop users from deleting customer records.
* Instead of deleting, the trigger logs the attempt.

### **Real-World Example 2: Audit Log Creation**

* Use **AFTER UPDATE** trigger to store old and new salary values when HR updates employee records.

* Automatically creates unique primary key values.

## **Question 19: Explain Transactions and How to Implement It**

A **transaction** is a group of operations executed as a single unit of work.
If any operation fails, the entire transaction rolls back.

### **ACID Properties of Transactions:**

* **A**tomicity → All or nothing
* **C**onsistency → Maintains database rules
* **I**solation → Transactions don’t affect each other
* **D**urability → Data is saved permanently after commit

### **How to implement a transaction in SQL:**

```sql
BEGIN TRANSACTION;
   -- SQL operations
IF @@ERROR <> 0
   ROLLBACK TRANSACTION;
ELSE
   COMMIT TRANSACTION;
```

---

### **Real-World Example 1: Banking Money Transfer**

Two operations must succeed:

1. Deduct from sender account
2. Add to receiver account

If any step fails → rollback.

### **Real-World Example 2: E-commerce Order Placement**

Operations include:

* Insert order
* Reduce stock
* Add payment entry
* Update wallet/coupon balance

If any fail → rollback.

## **Question 20: What Are Inner Joins?**

An **INNER JOIN** returns only matching records from both tables.

### **Definition:**

Returns rows where the join condition is true.

### **Syntax:**

```sql
SELECT * FROM A
INNER JOIN B
ON A.ID = B.ID;
```

### **Behavior:**

* Rows with matching values appear in results.
* Non-matching rows are excluded.

---

### **Real-World Example 1: Customers With Orders**

INNER JOIN Customers and Orders → returns only customers who have placed orders.

### **Real-World Example 2: Employees Assigned to Projects**

Joining Employees with ProjectAssignments shows only employees who are actually assigned to a project.

---


## **Question 21: Explain Left Join**

A **LEFT JOIN** returns all rows from the left table and only the matching rows from the right table. If there is no match, NULL values are returned for the right side.

### **Syntax:**

```sql
SELECT *
FROM TableA A
LEFT JOIN TableB B ON A.ID = B.ID;
```

### **Key Points:**

* Returns **all rows from the left table**
* Missing matches from right table are filled with **NULL**

---

### **Real-World Example 1: Customers and Orders**

LEFT JOIN Customers with Orders → shows every customer, including those with **no orders**.

Useful for identifying inactive customers.

---

### **Real-World Example 2: Employees and Assignments**

LEFT JOIN Employees with ProjectAssignments → shows all employees, even if some are **not assigned to any project**.

---

## **Question 22: Explain Right Join**

A **RIGHT JOIN** returns all rows from the right table and only matching rows from the left table.

### **Syntax:**

```sql
SELECT *
FROM TableA A
RIGHT JOIN TableB B ON A.ID = B.ID;
```

### **Key Points:**

* Returns **all rows from the right table**
* Missing matches from left table are filled with NULL

---

### **Real-World Example 1: Orders and Customers**

RIGHT JOIN Orders with Customers → returns all orders even if customer data is missing (e.g., archived customers).

---

### **Real-World Example 2: Courses and Enrollments**

RIGHT JOIN Enrollments with Courses → retrieves all courses even if no one enrolled yet.

---

## **Question 23: Explain Full Outer Join**

A **FULL OUTER JOIN** returns:

* All matching rows
* All non-matching rows from both tables
* Missing values are filled with NULL

### **Syntax:**

```sql
SELECT *
FROM TableA A
FULL OUTER JOIN TableB B ON A.ID = B.ID;
```

### **Key Points:**

* Combines the output of LEFT JOIN + RIGHT JOIN
* Shows everything, matching or not

---

### **Real-World Example 1: Students and Sports Participants**

FULL OUTER JOIN Students with SportsTeamMembers → shows:

* Students who are in sports
* Students not in sports
* External participants not in the school list

---

### **Real-World Example 2: Inventory vs Sales**

FULL OUTER JOIN Inventory with Sales → shows:

* Products in inventory but never sold
* Products that were sold but not currently in inventory

---

## **Question 24: Explain Cross Join**

A **CROSS JOIN** returns the **Cartesian product** of two tables. Every row in the first table is combined with every row in the second table.

### **Syntax:**

```sql
SELECT *
FROM TableA
CROSS JOIN TableB;
```

### **Key Points:**

* No join condition
* Returns maximum combinations
* Can produce very large result sets

---

### **Real-World Example 1: Product Variants**

If you have:

* 10 product models
* 5 color options

CROSS JOIN → 50 possible combinations.

Used in e-commerce catalogs.

---

### **Real-World Example 2: Timetable Generation**

If you CROSS JOIN:

* 7 days
* 6 time slots

You get 42 possible class schedule slots.

Useful for generating templates.

## **Question 27: Can We Have Unequal Columns in UNION?**

* **No.** The number of columns must be **the same** in each SELECT statement.
* Each column must have compatible data types.
* Otherwise, SQL will throw an error.

### **Real-World Example 1: Valid UNION**

```sql
SELECT Name, Salary FROM Employees
UNION
SELECT Name, Salary FROM Contractors;
```

* Both have 2 columns → valid

### **Real-World Example 2: Invalid UNION**

```sql
SELECT Name, Salary FROM Employees
UNION
SELECT Name FROM Contractors;
```

* Columns mismatch → error

---

## **Question 28: Can Columns Have Different Data Types in UNION?**

* **No.** Each corresponding column must have compatible data types.
* SQL automatically converts compatible types if possible (e.g., INT → FLOAT).
* If data types are incompatible (e.g., VARCHAR and DATE), it throws an error.

### **Real-World Example 1: Compatible Types**

* `SELECT Name, Age FROM Table1 UNION SELECT Name, CAST(Age AS INT) FROM Table2` → works because INT can be casted.

### **Real-World Example 2: Incompatible Types**

* `SELECT Name, DOB FROM Employees UNION SELECT Name, Salary FROM Contractors` → error (DATE vs FLOAT)

## **Question 30: When to Use GROUP BY?**

`GROUP BY` is used to group rows sharing a property, often used with aggregate functions.

### **Key Points:**

* Groups rows based on one or more columns
* Must be used when aggregate functions are applied on grouped columns

---

### **Real-World Example 1: Department Salary**

```sql
SELECT DepartmentID, AVG(Salary)
FROM Employees
GROUP BY DepartmentID;
```

* Calculates average salary per department

### **Real-World Example 2: Sales by Region**

```sql
SELECT Region, SUM(SalesAmount)
FROM Sales
GROUP BY Region;
```

* Total sales per region

---

## **Question 31: Can We Select Columns Not Part of GROUP BY?**

* **No.** Except if used inside an aggregate function
* Non-aggregated columns must appear in `GROUP BY`
* Selecting non-aggregated, non-grouped columns will throw an error

### **Real-World Example 1: Correct Usage**

```sql
SELECT DepartmentID, COUNT(EmployeeID)
FROM Employees
GROUP BY DepartmentID;
```

* EmployeeID is aggregated → valid

### **Real-World Example 2: Incorrect Usage**

```sql
SELECT DepartmentID, EmployeeName
FROM Employees
GROUP BY DepartmentID;
```

* EmployeeName not aggregated → error

## **Question 32: What Is HAVING Clause?**

The `HAVING` clause is used to filter groups created by `GROUP BY`, similar to how `WHERE` filters rows.

### **Key Points:**

* Filters aggregate results
* Cannot be used without `GROUP BY` (except in some databases)

### **Real-World Example 1: Departments with High Salary Average**

```sql
SELECT DepartmentID, AVG(Salary) AS AvgSalary
FROM Employees
GROUP BY DepartmentID
HAVING AVG(Salary) > 50000;
```

* Only departments with average salary > 50,000 appear

### **Real-World Example 2: Regions with High Sales**

```sql
SELECT Region, SUM(SalesAmount) AS TotalSales
FROM Sales
GROUP BY Region
HAVING SUM(SalesAmount) > 100000;
```

* Filters regions exceeding 100,000 in sales

---


## **Question 33: HAVING Clause vs WHERE Clause**

| Feature                          | WHERE | HAVING |
| -------------------------------- | ----- | ------ |
| Filters rows before aggregation  | ✅     | ❌      |
| Filters groups after aggregation | ❌     | ✅      |
| Can use aggregate functions      | ❌     | ✅      |

---

### **Real-World Example 1: Filtering Employees by Salary**

* WHERE: `SELECT * FROM Employees WHERE Salary > 50000;` → filters individual rows
* HAVING: `SELECT DepartmentID, AVG(Salary) FROM Employees GROUP BY DepartmentID HAVING AVG(Salary) > 50000;` → filters groups

### **Real-World Example 2: Sales by Region**

* WHERE filters: `SELECT * FROM Sales WHERE SalesAmount > 1000;` → rows with sales > 1000
* HAVING filters: `SELECT Region, SUM(SalesAmount) FROM Sales GROUP BY Region HAVING SUM(SalesAmount) > 100000;` → regions with total sales > 100,000

## **Question 38: How to Handle NULLs?**

NULL represents missing or unknown data. SQL provides several ways to handle NULL.

### **Techniques:**

* **IS NULL / IS NOT NULL** → filter rows
* **COALESCE()** → replace NULL with default value
* **IFNULL() / NVL()** → DB-specific functions

### **Real-World Example 1: Missing Emails**

* `SELECT * FROM Customers WHERE Email IS NULL;` → find customers without email
* `SELECT COALESCE(Email, 'Not Provided') FROM Customers;` → replace NULL

### **Real-World Example 2: Product Stock**

* NULL in `StockQuantity` → treat as 0
* `SELECT ProductName, COALESCE(StockQuantity, 0) FROM Products;`

## **Question 45: Explain SubQuery**

A **subquery** is a query nested inside another query, often used in SELECT, WHERE, or FROM clauses.

### **Purpose:**

* Break complex queries into smaller parts
* Filter results based on dynamic data

### **Real-World Example 1: Employee Salary**

* Find employees whose salary is higher than the average salary:

```sql
SELECT * FROM Employees
WHERE Salary > (SELECT AVG(Salary) FROM Employees);
```

### **Real-World Example 2: Customer Orders**

* Find customers who have placed orders above $1000:

```sql
SELECT * FROM Customers
WHERE CustomerID IN (SELECT CustomerID FROM Orders WHERE Amount > 1000);
```

## **Question 47: What Is Co-related Query?**

A **correlated subquery** refers to columns from the outer query. It runs once for each row of the outer query.

### **Real-World Example 1: Employees Earning Above Department Average**

```sql
SELECT E1.Name, E1.Salary
FROM Employees E1
WHERE E1.Salary > (SELECT AVG(E2.Salary) FROM Employees E2 WHERE E2.DepartmentID = E1.DepartmentID);
```

* Outer query depends on the inner query for each employee

### **Real-World Example 2: Orders Higher Than Customer Average**

* Find all orders for customers where order amount exceeds that customer’s average order value.

## **Question 50: Find Nth Highest Salary in SQL**

* Using **ROW_NUMBER()**:

```sql
SELECT * FROM (
   SELECT Salary, ROW_NUMBER() OVER (ORDER BY Salary DESC) AS RowNum
   FROM Employees
) AS Temp
WHERE RowNum = N;
```

* Replace `N` with the desired position (e.g., 3 for 3rd highest)

### **Real-World Example 1: 3rd Highest Paid Employee**

* Find which employee earns the 3rd highest salary to determine compensation hierarchy

### **Real-World Example 2: Bonus Distribution**

## **Question 51: Select Top Nth Highest Salary Using Correlated Queries**
